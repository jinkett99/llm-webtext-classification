### "bert-base-uncased"
- **More info:** https://huggingface.co/google-bert/bert-base-uncased
- **Summary:** bert-base-uncased is pre-trained for two objectives - Masked language modelling and next sentence prediction. Intended use is to be fine-tuned on a downstream task like sequence classification.

```python
#@title parse tokens into BERT model

model = BertForSequenceClassification.from_pretrained("bert-base-uncased", num_labels=2)
tmp_outputs = model(**tokens)

print('inputs: ', tokenizer.decode(tokens['input_ids'][0]))
print('outputs: ', tmp_outputs)
print(f"output distribution over labels: {torch.softmax(tmp_outputs.logits, dim=1)}")
```

```python
#@title Run pre-training - dataset preparation.

# 1. Instantiate the tokenizer (using bert-base-uncased as an example)
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# 2. Define the custom dataset class
class BertDataset(Dataset):
    def __init__(self, dataframe, tokenizer, max_length=512):
        # reset index to ensure indices are contiguous after splitting
        self.dataframe = dataframe.reset_index(drop=True)
        self.tokenizer = tokenizer
        self.max_length = max_length

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, idx):
        row = self.dataframe.iloc[idx]
        text = row['text_summary'] #Modified*
        label = row['label']
        return {"text": text, "label": label}

# 3. Custom collate_fn for batch tokenization
def collate_fn(batch):
    texts = [item['text'] for item in batch]
    labels = [item['label'] for item in batch]
    # Tokenize the batch of texts with dynamic padding
    encoding = tokenizer(
        texts,
        return_tensors='pt',
        padding=True,
        truncation=True,
        max_length=512
    )
    # Add labels to the encoding
    encoding['labels'] = torch.tensor(labels, dtype=torch.long)
    return encoding

# 4. Split the DataFrame into train and validation sets (80:20 split)
train_df, val_df = train_test_split(
    data,
    test_size=0.2,
    random_state=42,
    stratify=data['label']  # stratified split
)

# 5. Create dataset instances for train and validation
train_dataset = BertDataset(train_df, tokenizer)
val_dataset = BertDataset(val_df, tokenizer)
```

```python
#@title model training and hyperparameter tuning

# Define hyperparameter search space
learning_rate_options = [1e-5, 2e-5, 3e-5, 5e-5]
batch_size_options = [8, 16, 32]

# Perform random search
num_trials = 3 # 1-trial, no tuning.

# Track best trial metrics
best_hyperparams = {}
best_val_accuracy = 0  # For selection of best trial accuracy
best_metrics = {}

# Set device to GPU if available
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

for trial in range(num_trials):
    # Initialize trial metrics
    trial_metrics = []

    # Randomly select hyperparameters
    learning_rate = random.choice(learning_rate_options)
    batch_size = random.choice(batch_size_options)

    # Prepare Dataloaders - for training and validation
    train_dataloader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, collate_fn=collate_fn)
    val_dataloader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False, collate_fn=collate_fn)

    # Load the model - RE-TRAIN on "better" model.
    model = BertForSequenceClassification.from_pretrained("bert-base-uncased")
    model.to(device)

    # Define the optimizer and learning rate scheduler
    optimizer = AdamW(model.parameters(), lr=learning_rate)
    total_steps = len(train_dataloader) * 10  # 10 epochs
    scheduler = get_linear_schedule_with_warmup(optimizer, num_warmup_steps=0, num_training_steps=total_steps)

    # Initialize gradient scaler for mixed precision training
    scaler = torch.cuda.amp.GradScaler()

    # Training loop
    model.train()
    for epoch in range(10):  # Number of epochs ###
        total_train_loss = 0
        for batch in tqdm(train_dataloader, desc=f"Training Epoch {epoch+1}"):
            optimizer.zero_grad()

            input_ids = batch['input_ids'].to(device)
            attention_mask = batch['attention_mask'].to(device)
            token_type_ids = batch.get('token_type_ids')
            if token_type_ids is not None:
                token_type_ids = batch['token_type_ids'].to(device)
            labels = batch['labels'].to(device)

            # Use autocast for mixed precision
            with torch.cuda.amp.autocast():
                outputs = model(
                    input_ids=input_ids,
                    attention_mask=attention_mask,
                    token_type_ids=token_type_ids,
                    labels=labels
                )
                loss = outputs.loss

            total_train_loss += loss.item()

            # Scale loss and backpropagate
            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()
            scheduler.step()

        # Optionally clear the CUDA cache after each epoch
        torch.cuda.empty_cache()

    # Validation loop
    model.eval()
    val_labels = []
    val_preds = []
    val_probs = [] # for ROCAUC computation
    total_val_loss = 0

    with torch.no_grad():
        for batch in tqdm(val_dataloader, desc="Validation"):
            # Move batch data to device
            input_ids = batch['input_ids'].to(device)
            attention_mask = batch['attention_mask'].to(device)
            token_type_ids = batch.get('token_type_ids')
            if token_type_ids is not None:
                token_type_ids = token_type_ids.to(device)
            labels = batch['labels'].to(device)

            with torch.cuda.amp.autocast():
                outputs = model(
                    input_ids=input_ids,
                    attention_mask=attention_mask,
                    token_type_ids=token_type_ids,
                    labels=labels
                )
                loss = outputs.loss
            total_val_loss += loss.item()

            logits = outputs.logits
            probabilities = torch.softmax(logits, dim=-1)[:, 1]  # Get probability of class 1
            predictions = torch.argmax(logits, dim=-1)
            val_labels.extend(labels.cpu().numpy())
            val_preds.extend(predictions.cpu().numpy())
            val_probs.extend(probabilities.cpu().numpy())

    # Compute metrics
    avg_val_loss = total_val_loss / len(val_dataloader)
    val_accuracy = accuracy_score(val_labels, val_preds)
    val_precision = precision_score(val_labels, val_preds)
    val_recall = recall_score(val_labels, val_preds)
    val_f1 = f1_score(val_labels, val_preds)
    val_roc_auc = roc_auc_score(val_labels, val_probs)  # Compute ROC AUC

    # Store metrics for analysis
    trial_metrics.append({
        'epoch': epoch + 1,
        'val_loss': avg_val_loss,
        'val_accuracy': val_accuracy,
        'val_precision': val_precision,
        'val_recall': val_recall,
        'val_f1': val_f1,
        'val_roc_auc': val_roc_auc
    })

    print(f"Trial {trial+1}: Learning Rate: {learning_rate}, Batch Size: {batch_size}, Validation Accuracy: {val_accuracy}, Precision: {val_precision}, Recall: {val_recall}, F1-score: {val_f1}, ROC AUC: {val_roc_auc}")

    # Update the best hyperparameters if current trial is better
    if val_accuracy > best_val_accuracy:
        best_val_accuracy = val_accuracy
        best_hyperparams = {
            'learning_rate': learning_rate,
            'batch_size': batch_size
        }

        # Save the best model and tokenizer
        model.save_pretrained('models/BERTv2')
        tokenizer.save_pretrained('models/BERTv2')
        print("Saved new best model and tokenizer.")

# Modify to include: other metrics with best validation accuracy from trials (of 10 epochs)
print(f"Best Hyperparameters: {best_hyperparams}, Best Validation Accuracy: {best_val_accuracy}")
```
